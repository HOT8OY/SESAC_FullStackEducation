```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.rcParams["axes.unicode_minus"] = False
plt.rc("font", family="Malgun Gothic")
```

### 1. DataFrame객체 생성

```python
data={
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age" : [25, 30, 35, 40, 28],
    "City" : ["New York", "Los Angeles", "Chicago", "Houston", "San Francisco"],
    "Salary" : [70000, 80000, 120000, 90000, 75000]
}
df=pd.DataFrame(data)
df
```

### 2. 열과 행 선택

```python
# - `Name`과 `City` 열을 선택하여 출력하세요.
df[["Name","City"]]
```

```python
# 나이가 30 이상인 행을 선택하여 출력하세요.
df[df["Age"]>=30]
```

### 3. 행과 열 추가

```python
# `Country` 열을 추가하고, 모두 'USA'로 설정하세요.
df["Country"]=pd.DataFrame({"Country":["USA"]*len(df)})
df
```

```python
# 새로운 행을 추가하세요. (Name: Frank, Age: 33, City: Seattle, Salary: 85000)
exdata={"Name": "Frank", "Age": 33, "City": "Seattle", "Salary": 85000}
add_df=pd.DataFrame([exdata])
add_df
df = pd.concat([df,add_df], axis=0, ignore_index=True)
df
```

### 4. 데이터 삭제

```python
# `Salary` 열을 삭제하세요.
df.drop(["Salary"], axis=1)
```

```python
# 'David'가 있는 행을 삭제하세요.
df=df.drop(df[df["Name"]=="David"].index)
df
```

### 5. 데이터 수정

```python
# Age가 35 이상인 사람들의 Salary를 5% 증가시키세요.

df.loc[df["Age"]>=35, 'Salary'] = df.loc[df["Age"]>=35, 'Salary'] * 1.05
df
```

```python
# 'Alice'의 City를 'Boston'으로 수정하세요.
df.loc[0,'City']='Boston'
df
```

### 6. 데이터 타입 변환

```python
# Age와 Salary의 데이터 타입을 float로 변환하세요.
df[["Age","Salary"]]=df[["Age","Salary"]].astype(float)
df.info()
```
