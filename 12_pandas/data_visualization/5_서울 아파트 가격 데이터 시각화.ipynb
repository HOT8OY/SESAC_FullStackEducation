# **서울 아파트 실거래 데이터 시각화**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(
    style="darkgrid",
    palette="bright"
)

plt.rcParams['axes.unicode_minus'] = False
plt.rc("font", family = "Malgun Gothic")
```

```python
df= pd.read_csv("seoul_apart_02.csv")
df.head()
```

### 전체 데이터의 거래 금액을 히스토그램으로 그려보세요
* sns.histplot() 사용
* 인자는 `data`, `bins`만 사용
* `bins`는 30으로 설정

```python
sns.histplot(data=df["거래금액(만원)"], bins=30)
plt.title("전체 거래 금액 히스토그램")
plt.ylabel("거래 금액(만원)")
```

```python
# 월별 부동산 거래의 평당금액 평균값 확인
df.groupby("계약월")["평당금액"].mean()
```

```python
mon_price = df.groupby("계약월")[["평당금액"]].mean()
mon_price
```

```python
plt.plot(mon_price["평당금액"])
plt.show()
```

### 월별 아파트 실거래 계약건수를 막대그래프로 그려보세요
* sns.countplot()을 활용

```python
sns.countplot(data=df, x="계약월", hue="계약월", legend=False, palette="Set2")
plt.title("월별 아파트 실거래 계약건수")
plt.ylabel("계약건수")
plt.ylim(400,2000)
plt.grid(True, axis="y")
plt.show ()
```

```python
# 4월에 거래된 매물의 평당금액의 분포를 히스토그램으로 그리기
sns.histplot(data=df[df["계약월"]==4]["평당금액"])
```

```python
# 지역(구)별 거래금액을 비교하기 위한 막대그래프
plt.figure(figsize=(15,8))
plt.title("지역구별 평균 평당금액")
sns.barplot(data=df, x="구", y="평당금액")
plt.xticks(rotation=45)
plt.ylabel("평당금액(단위: 만원)")
```

```python
df.head()
```

```python
# 파이차트로 유형 비율 확인
type_ratio=df["유형"].value_counts()
type_ratio
plt.pie(x=type_ratio, labels=type_ratio.index, autopct="%1.1f%%")
plt.show()
```

### 지역구별 평당금액을 박스플롯으로 그리기
* sns.boxplot()
* 인자는 data, x, y를 사용

```python
# 지역구별 평당금액을 박스플롯으로 그리기
plt.figure(figsize=(12,5))
plt.xticks(rotation=45)
sns.boxplot(data=df, x="구", y="평당금액", hue="구", palette="Set2")
plt.title("지역구별 평당금액")
plt.ylabel("평당금액(단위: 만원)")
plt.grid(axis="y")
plt.show()
```

### "전용면적(평)" 컬럼과 "거래금액(만원)" 컬럼의 상관관계를 산점도로 그려보세요.

```python
sns.scatterplot(data=df, x="전용면적(평)", y="거래금액(만원)", hue="전용면적(평)", legend=False)
plt.title("거래금액(만원) | 전용면적(평) 산점도")
```

### 전체 수치형 데이터끼리의 상관계수를 히트맵으로 표현

```python
df_corr = df.corr(numeric_only=True)
df_corr
sns.heatmap(data=df_corr, annot=True) # annot=True -> 각 칸의 상관계수를 표시해줌
plt.title("상관계수 히트맵")
```
