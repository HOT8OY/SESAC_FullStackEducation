# **박스 그래프**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["axes.unicode_minus"] = False
plt.rc("font", family="Malgun Gothic")
```

```python
df=pd.read_csv("seoul_park.csv")
df.head()
```

```python
# 2018년 4월의 데이터 추출
df_2018_apr = df[(df["연"]==2018) & (df["월"]==4)]
df_2018_apr.head()
```

### Matplotlib를 이용한 boxplot
boxplot()

```python
# 2018년 4월 어린이 입장객 수의 boxplot
plt.boxplot(df_2018_apr["어린이"])
plt.show()
# 박스 내의 노란색 선은 중앙값
# 박스의 상단과 하단은 상위 25%(Q3), 하위 25%(Q1) 사분위수의 1, 3분위를 나타냄.
# O 표시는 이상치 데이터. 비정상적이거나 특별한 이유가 있는 데이터임.
```

```python
plt.boxplot([df_2018_apr["어린이"], df_2018_apr["어른"], df_2018_apr["청소년"]])
plt.xticks([1,2,3], labels=["어린이", "어른", "청소년"])
plt.grid(axis='y')
plt.title("2018년 4월의 입장객 수")
```

### Seaborn을 이용한 boxplot
sns.boxplot()

```python
# 2018년 4월 어린이 입장객 수의 boxplot
sns.boxplot(data=df_2018_apr, y="어린이")
```

```python
# 2018년 월별 어린이 입장객 수 시각화
df_2018=df[df["연"]==2018]
df_2018.head()

sns.boxplot(data=df_2018, x="월", y="어린이", palette="husl")
```

```python
# 2016년 월별 어른의 입장객 수를 박스 그래프로 그리세요.
df_2016 = df[df["연"]==2016]
sns.boxplot(data=df_2016, x="월", y="어른", palette="Set2")
```
