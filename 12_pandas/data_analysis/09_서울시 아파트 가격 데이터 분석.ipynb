# **서울 아파트 가격 데이터 분석**

```python
import pandas as pd
df=pd.read_csv("./data/seoul_apart_2022.csv")
df.head()
```

```python
df.info()
```

```python
# 사용하지 않는 "해제사유발생일", "중개사소재지", "번지", "본번", "부번", "도로명", "거래유형" 컬럼 삭제하기. 삭제가 원본 df에 적용되어야 한다.
df=df.drop(["해제사유발생일", "중개사소재지", "번지", "본번", "부번", "도로명", "거래유형"], axis=1)
df
```

```python
# 보기 쉽게 컬럼명을 변경
df=df.rename(columns={"전용면적(㎡)":"전용면적"})
df
```

```python
# '시군구' 컬럼의 주소를 활용하기 쉽게 '구' 컬럼과 '동' 컬럼으로 분리
df["구"]=df["시군구"].apply(lambda e: e.split()[1])
df["동"]=df["시군구"].apply(lambda e: e.split()[2])
df.head()
```

```python
# 전용면적이 60이하면 소형, 60보다 크고 85이하면 중형, 85보다 크고 102이하면 중대형, 102보다 크면 대형으로 분류하여 유형 컬럼을 생성
# 면적에 따라 아파트 유형을 분류하는 함수
def category(e):
    if e<=60:
        return "소형"
    elif e<=85:
        return "중형"
    elif e<=102:
        return "중대형"
    else:
        return "대형"
# .apply를 활용하여 적용된 유형 컬럼 생성
df["유형"]=df["전용면적"].apply(category)
df.head()
```

```python
# "계약년월" 컬럼과 "계약일" 컬럼을 합치고, 날짜타입으로 변경 (주의사항: "계약년월" 컬럼과 "계약일" 컬럼의 데이터 타입이 int 인 상태이다.)
df["계약일"]=df["계약년월"].astype('str')+df["계약일"].astype('str')

df["계약일"].info()
```

```python
# "계약년월" 컬럼과 "계약일" 컬럼을 합치고, 날짜타입으로 변경 (주의사항: "계약년월" 컬럼과 "계약일" 컬럼의 데이터 타입이 int 인 상태이다.)
df["계약일"]=pd.to_datetime(df["계약일"], format='%Y%m%d')

df["계약일"].info()
```

```python
# 계약일 데이터를 활용하여 "계약월"과 "계약요일" 컬럼 생성
df["계약월"]=df["계약일"].dt.month
df["계약요일"]=df["계약일"].dt.dayofweek
# 요일은 0~6까지의 숫자로 나오기 때문에 dict와 map을 이용하여 변환해준다.
df["계약요일"]=df["계약요일"].map({0:'월', 1:'화', 2:'수', 3:'목', 4:'금', 5:'토', 6:'일'})

df.head()
```

```python
# 데이터에 결측지가 있으면 정확한 분석이 어렵기때문애 null값이 있는지를 확인
df.isnull().sum()
```

```python
# dropna를 활용하여 "거래금액(만원)" 컬럼에 결측지가 존재하는 아파트 거래 데이터를 삭제하세요.
# 인덱스도 초기화 해주세요.
df=df.dropna(subset='거래금액(만원)', ignore_index=True)
df.info()
```

```python
# "거래금액(만원)" 컬럼의 콤마(,)를 삭제하고 숫자 형태 데이터로 바꾸기
df["거래금액(만원)"]=df["거래금액(만원)"].str.replace(",","")
df["거래금액(만원)"]=pd.to_numeric(df["거래금액(만원)"])

df["거래금액(만원)"].info()
```

```python
# 거래금액을 기준으로 내림차순 정렬
df.sort_values("거래금액(만원)", ascending=False).head(10)
```

```python
print("거래금액 평균값(만원):",df["거래금액(만원)"].mean())
print("거래금액 중앙값(만원):",df["거래금액(만원)"].median())
```

```python
# 전용면적 단위는 제곱미터로 되어있다. 우리에게 익숙한 평 단위로 변경해보자(1평은 대략 3.3제곱미터이다.)
# round(,2)는 소수점 두번째 자리에서 반올림
df["전용면적"] = round(df["전용면적"]/3.3, 2)
df=df.rename(columns={"전용면적":"전용면적(평)"})

df.head()
```

```python
# "평당금액" 컬럼을 생성해보자
# 평당금액은 "거래금액(만원)"값을 "전용면적(평)"값으로 나눈 값이다.
df["평당금액"] = df["거래금액(만원)"]/df["전용면적(평)"]
df["평당금액"]= round(df["평당금액"], 2)
df
```

```python
# 평당금액을 기준으로 내림차순 정렬
df.sort_values("평당금액", ascending=False).head()
```

```python
# 각 구별 평당금액의 평균값 확인
df.groupby('구')[["평당금액"]].mean()
```

```python
# 서초구의 동 중 평당금액 평균이 2번째로 높은 동의 이름을 변수 ans1에 저장하세요.
ans1 = df[df['구']=='서초구'].groupby('동')[['평당금액']].mean().sort_values(by="평당금액", ascending=False).iloc[1]
ans1
```
