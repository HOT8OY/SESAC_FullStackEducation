```python
import pandas as pd
```

```python
# 데이터셋 1: 매출 데이터 (sales)
data = {
    '날짜': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02', '2023-01-03', '2023-01-03'],
    '지점': ['강남점', '홍대점', '강남점', '홍대점', '강남점', '홍대점'],
    '상품': ['아메리카노', '라떼', '라떼', '아메리카노', '아메리카노', '라떼'],
    '수량': [100, 50, 70, 80, 110, 60],
    '단가': ['4,500', '5,000', '5,000', '4,500', '4,500', '5,000'] # 문자열(콤마 포함) 상태
}
sales = pd.DataFrame(data)
sales
```

```python
# 데이터셋 2: 지점 정보 (branch_info)
branch_data = {
    '지점': ['강남점', '홍대점', '성수점'],
    '매니저': ['김철수', '이영희', '박민수'],
    '오픈일': ['2020-01-01', '2021-03-01', '2022-05-01']
}
branch_info = pd.DataFrame(branch_data)
branch_info
```

```python
# Q1. 날짜 데이터 변환 (시계열 처리)
# 현재 문자열(Object)로 되어 있는 '날짜' 컬럼을 판다스의 날짜/시간 타입(datetime)으로 변환하여 저장하세요.
# 그 후, '날짜' 컬럼에서 '월(month)' 정보만 추출하여 '월'이라는 새로운 컬럼을 만드세요. (dt 접근자 사용)
sales.info()
sales["날짜"]=sales["날짜"].str.replace("-","")
sales["날짜"]=pd.to_numeric(sales["날짜"])
sales
sales["날짜"]=pd.to_datetime(sales["날짜"], format='%Y%m%d')
sales["월"]=sales["날짜"].dt.month
sales
```

```python
# Q2. 피벗 테이블 생성 (pivot_table)
# '지점'을 행(index)으로, '상품'을 열(columns)로 설정하고,
# 값(values)에는 '매출액'의 합계(sum)를 보여주는 피벗 테이블을 만드세요.
# (단, 매출이 없는 경우(NaN)는 0으로 채우세요. hint: fill_value 옵션)
```

```python
# Q3. 데이터 정렬 (Sorting)
# 전체 데이터에서 '매출액'이 가장 높은 순서대로 데이터를 정렬(내림차순)하고,
# 상위 3개의 판매 기록만 출력하세요.
sales["수량"]=pd.to_numeric(sales["수량"])
sales["단가"]=sales["단가"].str.replace(",","")
sales["단가"]=pd.to_numeric(sales["단가"])
sales["매출액"]=sales["수량"]*sales["단가"]
sales.sort_values("매출액", axis=0, ascending=False).head(3)
```
