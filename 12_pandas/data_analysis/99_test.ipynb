```python
import pandas as pd
import numpy as np
```

```python
# 데이터셋 1: 매출 데이터 (sales)
data = {
    '날짜': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02', '2023-01-03', '2023-01-03'],
    '지점': ['강남점', '홍대점', '강남점', '홍대점', '강남점', '홍대점'],
    '상품': ['아메리카노', '라떼', '라떼', '아메리카노', '아메리카노', '라떼'],
    '수량': [100, 50, 70, 80, 110, 60],
    '단가': ['4,500', '5,000', '5,000', '4,500', '4,500', '5,000'] # 문자열(콤마 포함) 상태
}
sales = pd.DataFrame(data)
sales
```

```python
# 데이터셋 2: 지점 정보 (branch_info)
branch_data = {
    '지점': ['강남점', '홍대점', '성수점'],
    '매니저': ['김철수', '이영희', '박민수'],
    '오픈일': ['2020-01-01', '2021-03-01', '2022-05-01']
}
branch_info = pd.DataFrame(branch_data)
branch_info
```

```python
# Q1. 데이터 전처리 (형변환)
# sales 데이터프레임의 '단가' 컬럼은 현재 콤마(,)가 포함된 문자열(object) 형태입니다.

# 콤마를 제거하고, 정수형(int 또는 int64)으로 변환하여 다시 '단가' 컬럼에 저장하세요.
sales["단가"]=sales["단가"].str.replace(",","")
sales["단가"]=pd.to_numeric(sales["단가"])
sales
```

```python
# Q2. 파생 변수 생성 (연산)sales 데이터프레임에 **'매출액'**이라는 새로운 컬럼을 추가하세요.
# 계산 공식: 수량 * 단가
sales["매출액"]=sales["수량"]*sales["단가"]
sales
```

```python
# Q3. 데이터 추출 (조건 필터링)
# sales 데이터프레임에서 **'강남점'**에서 판매된 데이터 중, '매출액'이 450,000원 이상인 행(Row)만 추출하여 출력하세요.
sales[(sales["지점"] == "강남점") & (sales["매출액"] >= 450000)]
```

```python
# Q4. 데이터 요약 (Groupby)
# 지점별로 총 얼마를 팔았는지 알고 싶습니다. sales 데이터를 이용하여 **'지점'**별 '매출액'의 합계를 구하세요. (결과는 Series 혹은 DataFrame 형태 모두 무방합니다)
sales.groupby('지점')[['매출액']].sum()
```

```python
# Q5. 데이터 병합 (Merge)
# sales 데이터와 branch_info 데이터를 '지점' 컬럼을 기준으로 합치려고 합니다.
merged = pd.merge(sales, branch_info, how='left')
merged
# sales 데이터는 모두 유지되어야 합니다 (Left Join).

# 합친 결과에서 매니저 정보가 잘 들어왔는지 확인하기 위해 병합된 데이터프레임을 출력하세요.
```
