### 데이터 정제하기

```python
import pandas as pd

df=pd.read_csv("./data/seoul_park03.csv")
df.head()
```

### 데이터 정령: sort_values

```python
# sort_valuse("정렬할 컬럼 이름", 정렬 방법)
# 특정 컬럼의 값을 기준으로 전체 데이터를 정렬한다
df_sorted = df.sort_values("총계", ascending=False) # ascending 생략하면 기본값은 True(오름차순)

df_sorted.head(10)
```
