# Chat Completion

### 환경설정

```python
# 환경설정
from dotenv import load_dotenv

load_dotenv()
```

```python
import os

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
```

```python
from openai import OpenAI

client = OpenAI(api_key=OPENAI_API_KEY)

response = client.chat.completions.create( # API쪽으로 POST 요청을 보냄
    model="gpt-4o-mini",
    messages=[
        {"role":"system", "content":"당신은 친절하고, 상세한 설명을 잘하는 챗봇입니다."},
        {"role":"user", "content":"안녕 나의 이름은 판다야!!"},
        {"role":"assistant", "content":"안녕, 판다야! 만나서 반가워. 어떻게 도와줄까? 궁금한 것이나 이야기하고 싶은 것이 있으면 언제든지 말해줘!"},
        {"role":"user", "content":"내 이름이 뭐라고?"}
    ],
    temperature=1,
    max_completion_tokens=4096,
    top_p=1
)

print(response.choices[0].message.content)
```

### stream 처리

```python
client = OpenAI() # api_key를 작성하지 않으면, 자동으로 환경변수 OPENAI_API_KEY를 참조한다. 대신 load_dotenv()는 필수임. 다른 이름으로 저장했다면 적어줘야 함.

stream_response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"user","content":"스트림 테스트하는데, 긴 응답메세지를 보내주세요."}
    ],
    stream=True # 생성되는 단어(토큰)조각들을 실시간으로 계속해서 보내준다.
)

for chunk in stream_response:
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="") # end의 기본값은 end="\n"으로 단어가 끝날 때 마다 개행되지만, 줄바꿈 없이 계속 하도록 만듦.

print("🍗응답 완료!!🍗")
```

```python
import tiktoken

gpt35 = tiktoken.encoding_for_model("gpt-3.5-turbo")
gpt4o = tiktoken.encoding_for_model("gpt-4o")
```

```python
text = """
서울시버스조합은 이달 30일부터 '2026년 새해 희망버스 캠페인'도 진행한다. 소방관들이 작성한 새해 인사 글과 그림을 시내버스 외부에 부착하는 캠페인이다.

서울시버스조합은 작년에 이어 2년째 소방관 응원 행사를 이어가고 있다. 올해는 송파소방서를 비롯해 총 10개 소방서에 커피차 음료를 지원하고 다육이 액자 1천200개, 서울라면 400상자를 나눠줄 계획이다.
"""

gpt35_tokens = gpt35.encode(text)
gpt4o_tokens = gpt4o.encode(text)

print(len(gpt35_tokens))    # 214
print(len(gpt4o_tokens))    # 136
```
