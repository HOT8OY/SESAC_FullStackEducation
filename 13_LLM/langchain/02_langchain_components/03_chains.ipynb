### Simple Chain(기본 체인)

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser   # 출력 파서(문자열로 변환)

# PromptTemplate 생성
prompt = PromptTemplate(
    template="{country}의 수도는 어디인가요?",
    input_variables=['country']     # 사용자가 입력할 변수명 지정
)

# LLM Model 생성
model = ChatOpenAI(
    model="gpt-5-nano",
    temperature=0
)

# OutputParser 생성
output_parser = StrOutputParser()   # 모델 출력 결과에서 텍스트 내용만 뽑아주는 파서

# Chain 연결 (파이프(|) 연산자 사용)
chain = prompt | model | output_parser

result = chain.invoke(input={"country":"대한민국"})

print(result)
```

```python
# 체인 2 : 요약 및 연결
from langchain_core.runnables import RunnableSequence # 여러 Runnable(체인)을 순서대로 실행할 수 있음

# 요약 프롬프트 생성
sum_prompt = PromptTemplate(
    template="다음의 글을 짧게 요약하세요 : {text}",
    input_variables=['text']
)

# 요약 체인 생성
summary_chain = sum_prompt | llm

# 체인 연결
# 앞의 체인의 출력이 뒤의 체인 입력으로 자동 전달된다.
overall_chain = RunnableSequence(translation_chain, summary_chain)

final_output = overall_chain.invoke(input_text)
print(final_output)
```

```python
# 조건부 체인
from langchain_core.runnables import RunnableBranch, RunnablePassthrough

llm = ChatOpenAI(model='gpt-5-nano', temperature=0)

# 평가 체인
grading_prompt = PromptTemplate(
    template="당신은 냉철한 평가자입니다. 아래 답변을 1~5점으로 평하세주세요:\n\n{text}",
    input_variables=['text']
)

grading_chain = grading_prompt | llm

# 기본 체인
default_prompt = PromptTemplate(
    template="당신은 사용자의 질문에 답하는 친절한 챗봇입니다. \n\n{text}",
    input_variables=['text']
)

default_chain = default_prompt | llm

def grading_routing_fn(input_dict) -> bool:
    text = input_dict.get('text', '')
    return text.strip().startswith('평가')

# cond_chain = RunnableBranch(
#     (조건 함수, True일때 실행할 체인),
#     False일때 실행할 체인
# )
cond_chain = RunnableBranch(
    (grading_routing_fn, grading_chain),
    default_chain
)

# 일반 질문
print(cond_chain.invoke(input={'text':'인공지능이 무엇인가요?'}))
```

```python
# 평가 질문
evaluation_input="평가: 인공지능(AI)은 컴퓨터가 인간과 비슷하게 지능적으로 보이는 일을 수행하도록 만드는 기술 전체를 말합니다. 예를 들면 학습, 추론, 이해, 문제 해결, 언어 이해 같은 것들입니다.\n\n주요 특징과 구분\n- 좁은 AI(약칭 AI, Narrow AI): 특정 작업에 특화된 지능으로 현재 우리가 많이 쓰는 형태입니다. 예: 음성인식, 이미지 인식, 번역, 추천 시스템.\n- 일반 AI(강한 AI, General AI): 인간처럼 모든 지적 작업을 다 수행할 수 있는 수준의 AI로 아직 실현되지 않았습니다.\n\n작동 방식의 간단한 설명\n- 데이터와 알고리즘으로 학습: 예측이나 의사결정을 잘하도록 모델을 데이터를 통해 조정합니다.\n- 기계학습/딥러닝: 데이터에서 패턴을 자동으로 발견하는 대표적 방법들입니다. 딥러닝은 뇌의 신경망을 닮은 구조로 복잡한 패턴을 다루는 데 강합니다.\n- 강화학습: 환경과 상호작용해 보상 신호를 바탕으로 행동을 개선합니다.\n- 규칙 기반 시스템: 사람이 미리 만든 규칙으로 작동하는 방식도 과거에 많이 사용되었습니다.\n\n일상에서의 예\n- 스마트폰 음성비서, 번역 앱\n- 추천 알고리즘(영화/음악 추천)\n- 이메일 스팸 필터, 사진 속 인물 인식\n- 자율주행 차량, 로봇의 제어 시스템\n\n한계와 고려할 점\n- 데이터 편향성 문제: 학습 데이터의 편향이 결과에 반영될 수 있습니다.\n- 일반상식이나 맥락 이해의 한계: 상황에 따른 직관적 판단이 부족할 때가 많습니다.\n- 해석 가능성: 왜 그런 판단을 내렸는지 설명이 어렵기도 합니다.\n- 안전·윤리: 프라이버시, 악용 가능성, 책임 소재 등이 중요한 이슈입니다.\n\n필요하신 특정 분야나 더 자세한 설명이 있다면 말씀해 주세요. 예를 들어 AI가 어떻게 작동하는지, 실생활 사례를 더 자세히 알아보고 싶다거나, 특정 기술(딥러닝, 강화학습 등)에 초점을 맞추고 싶으신지 알려주시면 맞춤해 드리겠습니다."
print(cond_chain.invoke(input={'text': evaluation_input}))
```

```python
# RunnablePassthrough 활용
# {'text': RunnablePassthrough()} : 들어온 문자열을 그대로 'text' 키의 값으로 할당
pass_chain = {'text': RunnablePassthrough()} | cond_chain

print(pass_chain.invoke('평가: 코끼리는 어류이다.'))
```
