```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
```

```python
image_url = "https://k-life.co/files/attach/images/2023/10/19/b186671c1be4bf001c2cdbc9b8d18e72.jpg"

model = ChatOpenAI(model="gpt-4o", temperature=0)

# 멀티모달 메시지 구성
# 텍스트와 이미지 URL을 함께 리스트로 전달
message = HumanMessage(
    content=[
        {"type":"text", "text":"이 이미지는 무엇인가요? 자세히 묘사해주세요."}, # 질문
        {"type":"image_url", "image_url":{"url":image_url}}               # 이미지 데이터
    ]
)

response = model.invoke([message])
print(response.content)
```

```python
# 영수증 이미지
receipt_url = "https://img.etnews.com/photonews/1707/971120_20170705143932_354_0001.jpg"

prompt_text =  "이 이미지에서 상호명(Store Name)과 총 금액(Total Amount)을 찾아서 알려줘"

message = HumanMessage(
    content=[
        {"type":"text", "text":prompt_text}, # 질문
        {"type":"image_url", "image_url":{"url":receipt_url}}               # 이미지 데이터
    ]
)

response = model.invoke([message])
print(response.content)
```
